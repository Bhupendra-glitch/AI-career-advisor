```python
def binary_search(arr, x):
  low, high = 0, len(arr)-1
  while low <= high:
    mid = (low + high)//2
    if arr[mid] == x:
      return mid
    elif arr[mid] < x:
      low = mid + 1
    else:
      high = mid - 1
  return -1
```

```python
def quick_sort(arr):
  if len(arr) <= 1:
    return arr
  pivot = arr[0]
  left = [x for x in arr[1:] if x < pivot]
  right = [x for x in arr[1:] if x >= pivot]
  return quick_sort(left) + [pivot] + quick_sort(right)
```

```python
import heapq

def dijkstra(graph, start):
    dist = {node: float("inf") for node in graph}
    dist[start] = 0
    pq = [(0, start)]

    while pq:
        cur_dist, node = heapq.heappop(pq)
        if cur_dist > dist[node]:
            continue
        for neigh, weight in graph[node]:
            new_dist = cur_dist + weight
            if new_dist < dist[neigh]:
                dist[neigh] = new_dist
                heapq.heappush(pq, (new_dist, neigh))

    return dist

graph = {
    'A':[('B',1),('C',4)],
    'B':[('C',2),('D',5)],
    'C':[('D',1)],
    'D':[]
}
print(dijkstra(graph, 'A'))

```

```python
def lcs(a, b):
    dp = [[0]*(len(b)+1) for _ in range(len(a)+1)]
    for i in range(1, len(a)+1):
        for j in range(1, len(b)+1):
            if a[i-1] == b[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])
    return dp[-1][-1]

print(lcs("ABCBDAB", "BDCABC"))
```

```python
def is_palindrome(s):
    if len(s) <= 1:    # base case
        return True
    if s[0] != s[-1]:  # mismatch
        return False
    return is_palindrome(s[1:-1])  # recurse on inner substring

# Example
print(is_palindrome("madam"))   # True
print(is_palindrome("hello"))   # False
```

```python
def count_x(lst, x):
    if len(lst) == 0:      # base case
        return 0
    if lst[0] == x:
        return 1 + count_x(lst[1:], x)
    else:
        return count_x(lst[1:], x)

# Example
lst = [1, 2, 3, 2, 4, 2, 5]
print(count_x(lst, 2))   # Output: 3
```

```python
'''
Divide Number Into Almost Equal Parts
Write a function divide_into_almost_equal_parts(n: int, k: int) -> list that takes two integers, n and k, and creates a list of size k where the elements are approximately equal and sum up to n. The list should contain larger numbers towards the beginning.

The larger numbers should be prioritized for the earlier parts of the list, and you should ensure that the sum of the list matches n.

Constraints

The integers n (total) and k (parts) are both positive integers.
The output list will always be of length k.
NOTE: This is a function type question, you don't have to take input or print the output, just have to complete the required function definition.

Example

>>> divide_into_almost_equal_parts(5, 3)
[2, 2, 1]
>>> divide_into_almost_equal_parts(16, 3)
[6, 5, 5]
>>> divide_into_almost_equal_parts(12, 4)
[3, 3, 3, 3]
>>> divide_into_almost_equal_parts(10, 3)
[4, 3, 3]
'''
```

```python
def divide_into_almost_equal_parts(n: int, k: int) -> list:
  base = n//k
  remainder = n % k

  result = []
  for i in range(k):
    if i < remainder:
      result.append(base + 1)
    else:
      result.append(base)
  return result
print(divide_into_almost_equal_parts(5, 3))
print(divide_into_almost_equal_parts(16, 3))
```

```python
def unique_sum_pairs(nums: list, k: int) -> set:
  result = []
  seen = set()
  for num in nums:
    complement = k -num
    if complement in seen:
      result.add(tuple(sorted(num, complement)))
    seen.add(num)
  return result
```

```python
from logging import critical
def average(marks):
  return sum(marks)/len(marks)

def filter_student(data: dict, criteriat: str) -> set:

  if criteria == "excellent":
    return {
        name for name, marks in data.items()
        if average(marks) >= 90
    }
  elif criteria == "average":
    return {
        name for name, marks in data.items()
        if 60 <= average(marks) < 90
    }
  elif criteria == "poor":
    return {
        name for name, marks in data.items()
        if average(marks) < 60
    }
  else:
    critical("Invalid criteria")
    return set()
```

```python
'''
Section 2
Question 1: Sales Analysis by Product

Question

Given

a list of dictionaries
each dictionary has
product
sales list

Return only those products whose total sales are at least the threshold.

For every selected product return

total
count
average
max'''


def sales_analysis(data, threshold):
  result = {}

  for item in data:
    product = item["product"]
    sales = item["sales"]

    total = sum(sales)
    if total >= threshold:
      result[product] = {
          "total": total,
          "count": len(sales),
          "average": total/len(sales),
          "max": max(sales)
      }
    return result
```

```python
'''
Question 2 (10 Marks): Encode by Shifting Ends
Question

Write a function encode_by_shifting_ends(s).

Rules:

Shift the first character forward by 1 in ASCII.
Shift the last character backward by 1 in ASCII.
Keep middle characters unchanged.
If the first character is 'z', leave it unchanged.
If the last character is 'a', leave it unchanged.

Examples:

encode_by_shifting_ends("abc")   -> "bbb"
encode_by_shifting_ends("hello") -> "ielln"
encode_by_shifting_ends("zoo")   -> "zon"
encode_by_shifting_ends("a")     -> "a"
encode_by_shifting_ends("za")    -> "za"'''

def encode_by_shifting_ends(s: str) -> str:
  if len(s) == 1:
    return s
  first = s[0]
  last = s[-1]

  if first != 'z':
    first = chr(ord(first) + 1)
  if last != 'a':
    last = chr(ord(last) - 1)
  return first + s[1:-1] + last
```

```python
'''Q3. Playing Card to Tuple
A card is represented by a string such as "AH", "10D" or "KS".

Write a function card_value(card) that converts a card into a tuple:
- First element: numerical value (A = 1, J = 11, Q = 12, K = 13, numeric cards retain their number).
- Second element: suit (H = Hearts, D = Diamonds, C = Clubs, S = Spades).'''


def card_value(card):
  rank = card[:-1]
  suit = card[-1]

  ranks = {
      "A": 1,
      "J": 11,
      "Q": 12,
      "K": 13
  }
  suits = {
      "H": "hearts",
      "D": "Diamonds",
      "C": "Clubs",
      "S": "Spades"
  }

  if rank in ranks:
    value = ranks[rank]
  else:
    value = int(rank)

  return (value, suits[suit])
print(card_value("AH"))
print(card_value("10D"))
print(card_value("KS"))
```
